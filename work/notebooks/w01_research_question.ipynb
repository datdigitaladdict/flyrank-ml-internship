Three numbers from the starter dataset show why this lane is worth pursuing: how much of the inventory is currently declining, how narrow a simple rule-based filter is, and how much a trained model can improve targeting over a hand-written rule.

```python
import pandas as pd, json

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

declining_rate = df["is_declining_label"].mean()
print(f"Share of pages currently labeled 'declining': {declining_rate:.3f}")

stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print(f"Pages meeting 'stale + visible' criteria: {len(stale_visible)} of {len(df)}")

res = json.load(open("outputs/model_results.json"))
base = res["baseline"]["baseline_precision_at_50"]
rf = res["models"]["random_forest"]["precision_at_50"]
print(f"Hand-rule Precision@50: {base:.3f}   Random forest Precision@50: {rf:.3f}")
```

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

These numbers show real patterns, not proof of cause. 54.2% of pages are labeled "declining," but only 17 of 30,000 pages meet a simple stale-and-visible rule, showing that a fixed rule alone is far too narrow to be useful at scale. The random forest reached 0.740 Precision@50 versus the hand rule's 0.240, showing a learned ranking can meaningfully outperform a fixed rule on this starter slice. This does not mean refreshing a page will cause it to recover, that would require an actual before/after experiment. It also doesn't rule out other explanations for decline, like consolidation or seasonality, which I have not checked yet.
