## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can only combine a handful of conditions a human can hold in their head, for example days_since_last_update >= 180 and impressions_90d >= 500. In my earlier notebook, that exact rule only flagged 17 of 30,000 pages, far too narrow to be useful at scale. A model can weigh dozens of signals simultaneously (impressions, position, CTR, word count, age, and their interactions) and find patterns too messy to write as an if-statement. This is exactly why the random forest reached 0.740 Precision@50 versus the hand rule's 0.240 in my earlier pipeline run, the pattern of what makes a page worth reviewing isn't captured by a few thresholds alone.
